# Parts of Speech Tagging 

```python
# import SaCy library
import spacy
# Load the English NLP model
nlp = spacy.load("en_core_web_sm") 
```

```python
doc = nlp(u"Apple is looking at buying U.K. startup for $1 billion")
```

```python
doc.text
```

```python
doc[0]
```

```python
doc[0].pos_  # part of speech tag
```

```python
doc[0].tag_  # detailed part of speech tag
```

```python
spacy.explain('NNP')
```

```python
for word in doc:
    print(f'{word.text:{12}} {word.pos_:{6}} {word.tag_:{6}} {spacy.explain(word.tag_)}')
```

```python
doc2 = nlp(u"Autonomous cars shift insurance liability toward manufacturers") 
for word in doc2:
    print(f'{word.text:{12}} {word.pos_:{6}} {word.tag_:{6}} {spacy.explain(word.tag_)}')
```

```python
doc3 = nlp(u"AI is the future, and the future is here.")
for word in doc3:
    print(f'{word.text:{12}} {word.pos_:{6}} {word.tag_:{6}} {spacy.explain(word.tag_)}')
```

```python
doc4 = nlp(u"Can you believe that NASA was established in 1958? on July 29.")
for word in doc4:
    print(f'{word.text:{12}} {word.pos_:{6}} {word.tag_:{6}} {spacy.explain(word.tag_)}')
```

```python
from spacy import displacy
displacy.render(doc4,style='dep',jupyter=True)
```

```python
options={
    'distance':96,
    'compact':True,
    'color':'#fff',
    'bg':'#00a65a'
}
displacy.render(doc4,style='dep',jupyter=True,options=options)
```
