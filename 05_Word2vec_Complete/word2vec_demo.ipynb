```python
from gensim.models import KeyedVectors

path = "/content/word2vec_model/GoogleNews-vectors-negative300.bin.gz"
word2vec = KeyedVectors.load_word2vec_format(path, binary=True)

print("Model loaded successfully!")
print(f"Vector size: {word2vec.vector_size}")
```

```python
word2vec['man']
```

```python
word2vec.most_similar("computer")
```

```python
word2vec.most_similar("AI")
```

```python
word2vec.most_similar("Boy")
```

```python
vec = word2vec['king'] - word2vec['man'] + word2vec['woman']
word2vec.most_similar([vec])
```

```python
vec = word2vec['PKR'] - word2vec['Pakistan'] + word2vec['EUROPE']
word2vec.most_similar([vec])
```
