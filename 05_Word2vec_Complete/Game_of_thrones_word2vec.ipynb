```python
import numpy as np
import pandas as pd
import gensim
import os
from nltk.tokenize import sent_tokenize
from gensim.utils import simple_preprocess
```

```python
story = []
for filename in os.listdir('data/'):
  # Try different encodings if 'latin-1' doesn't work for all files
  try:
      f = open(os.path.join('data/',filename), encoding='latin-1')
      corpus = f.read()
      raw_sent = sent_tokenize(corpus)
      for sent in raw_sent:
        story.append(simple_preprocess(sent))
  except UnicodeDecodeError:
      print(f"Could not decode file {filename} with latin-1 encoding. You might need to try a different encoding.")
```

```python
story[0:5]
```

```python
model = gensim.models.Word2Vec(
    window=10,
    min_count=2,
)
```

```python
model.build_vocab(story)
```

```python
model.train(story,total_examples=model.corpus_count,epochs=model.epochs)
```
